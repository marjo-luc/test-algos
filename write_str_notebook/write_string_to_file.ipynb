```python
text = "default string"
output_path = "output.txt"
```

```python
import os

OUTPUT_DIR = "output"
os.makedirs(OUTPUT_DIR, exist_ok=True)
full_output_path = os.path.join(OUTPUT_DIR, output_path)

with open(full_output_path, "w") as f:
    f.write(text)

print(f"Wrote {len(text)} characters to {full_output_path}")
```
